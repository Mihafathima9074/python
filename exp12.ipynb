```python
#Length of the longest word

def longest_word(words):
    lengths = [len(word) for word in words]
    return max(lengths)

words = input("Enter words separated by spaces: ").split()

print("Length of longest word:", longest_word(words))
```

```python
# Pattern Printing Using Function

def pattern(n):
    for i in range(1, n + 1):
        print("* " * i)

    for i in range(n - 1, 0, -1):
        print("* " * i)

n = int(input("Enter the number of rows: "))
pattern(n)
```

```python
# Generate All factors of a number

def factors(n):
    result = [i for i in range(1, n + 1) if n % i == 0]
    return result

n = int(input("Enter a number: "))
print("Factors:", factors(n))
```

```python
#area of square,rectanle and triangle using lambda functions
square = lambda side: side * side
rectangle = lambda length, breadth: length * breadth
triangle = lambda base, height: 0.5 * base * height

side = float(input("Enter the side of the square: "))
print("Area of square =", square(side))

length = float(input("Enter the length of the rectangle: "))
breadth = float(input("Enter the breadth of the rectangle: "))
print("Area of rectangle =", rectangle(length, breadth))

base = float(input("Enter the base of the triangle: "))
height = float(input("Enter the height of the triangle: "))
print("Area of triangle =", triangle(base, height))
```
